## Conclusion & principaux enseignements


L’analyse de la qualité des données a permis d’identifier plusieurs valeurs manquantes dans les variables Nucléaire, Stockage batterie, Déstockage batterie, Eolien terrestre et Eolien offshore. Toutefois, aucun NaN n’a été détecté entre la première et la dernière observation disponible de ces variables. Les valeurs manquantes sont donc principalement structurelles, liées à la période de disponibilité des données, plutôt qu’à des anomalies ponctuelles dans la série temporelle.
Plusieurs opérations de nettoyage ont ensuite été réalisées :
- les représentations textuelles des valeurs manquantes telles que ND, N/A, -, etc. ont été vérifiées afin de s’assurer qu’elles sont correctement interprétées comme des NaN lors de l’ingestion ;
- les variables quantitatives ont été converties en type float afin de garantir leur exploitation dans les analyses statistiques ;
- les colonnes Date et Heures ont été supprimées après leur fusion dans une variable unique datetime, évitant ainsi la redondance ;
- Nature a été retirée car elle n’apportait pas d’information utile pour l’analyse retenue ;
- Stockage batterie et Déstockage batterie ont été retirées en raison de leur couverture temporelle limitée par rapport à la période globale étudiée ;
- Nucléaire a également été supprimée : les observations disponibles pour la Bretagne sont exclusivement égales à 0, la variable n’apportant donc aucune variabilité exploitable pour l’analyse.
Au terme de cette étape, le dataset obtenu présente une structure plus cohérente, des types adaptés aux traitements statistiques et une gestion explicite des valeurs manquantes. Il constitue ainsi une base plus fiable pour les prochaines étapes d’analyse exploratoire et de comparaison des dynamiques énergétiques.
