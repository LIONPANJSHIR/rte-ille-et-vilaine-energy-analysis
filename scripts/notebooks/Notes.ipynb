### Techniques utiles a connaitre

* 3 différentes  methôdes de comparaison de la dispersion

    | Méthode                       | Ce qu'elle mesure                             | Adaptée à ton objectif ?                    |
    | ----------------------------- | --------------------------------------------- | ------------------------------------------- |
    | Écart-type                    | Dispersion des consommations annuelles en TWh | Oui, mais dépend de l'échelle               |
    | Normalisation base 100        | Évolution relative depuis 2013                | Oui, pour comparer les trajectoires         |
    | Coefficient de variation (CV) | Dispersion relativement à la moyenne          | Le plus adapté pour comparer la variabilité |

    Le coefficient de variation est défini par :

    $$ CV=\frac{\text{Écart-type}}{\text{Moyenne}}\times100 $$

    Par exemple, un CV de 5 % signifie que l'écart-type représente 5 % de la moyenne.


    Un CV plus élevé indique une plus forte dispersion relative des niveaux annuels, mais ne prouve pas que les variations d'une année à l'autre sont plus brutales.
